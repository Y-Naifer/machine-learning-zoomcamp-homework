```python
def split(df, seed):
    n = len(df)
    n_val = int(n * 0.2)
    n_test = int(n * 0.2)
    n_train = n - n_val - n_test

    np.random.seed(seed)
    idx = np.arange(n)
    np.random.shuffle(idx)

    df_train = df.iloc[idx[:n_train]].reset_index(drop=True)
    df_val = df.iloc[idx[n_train:n_train + n_val]].reset_index(drop=True)
    df_test = df.iloc[idx[n_train + n_val:]].reset_index(drop=True)
    return df_train, df_val, df_test
```

```python
def train_linear_regression(X, y):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])
    XTX = X.T.dot(X)
    XTX_inv = np.linalg.inv(XTX)
    w = XTX_inv.dot(X.T).dot(y)
    return w[0], w[1:]

def train_linear_regression_reg(X, y, r=0.0):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])
    XTX = X.T.dot(X)
    XTX = XTX + r * np.eye(XTX.shape[0])
    XTX_inv = np.linalg.inv(XTX)
    w = XTX_inv.dot(X.T).dot(y)
    return w[0], w[1:]

def rmse(y, y_pred):
    se = (y - y_pred) ** 2
    return np.sqrt(se.mean())
```

```python
df_train, df_val, df_test = split(df, 42)
feat = ['engine_displacement', 'horsepower', 'vehicle_weight', 'model_year']

# Option A: fill with 0
train_0, val_0 = df_train.copy(), df_val.copy()
train_0['horsepower'] = train_0['horsepower'].fillna(0)
val_0['horsepower'] = val_0['horsepower'].fillna(0)

X_train, y_train = train_0[feat].values, train_0['fuel_efficiency_mpg'].values
X_val, y_val = val_0[feat].values, val_0['fuel_efficiency_mpg'].values

w0, w = train_linear_regression(X_train, y_train)
y_pred = w0 + X_val.dot(w)
print("fill 0:", round(rmse(y_val, y_pred), 3))

# Option B: fill with mean (computed on train only)
mean_hp = df_train['horsepower'].mean()
train_m, val_m = df_train.copy(), df_val.copy()
train_m['horsepower'] = train_m['horsepower'].fillna(mean_hp)
val_m['horsepower'] = val_m['horsepower'].fillna(mean_hp)

X_train_m, y_train_m = train_m[feat].values, train_m['fuel_efficiency_mpg'].values
X_val_m, y_val_m = val_m[feat].values, val_m['fuel_efficiency_mpg'].values

w0m, wm = train_linear_regression(X_train_m, y_train_m)
y_pred_m = w0m + X_val_m.dot(wm)
print("fill mean:", round(rmse(y_val_m, y_pred_m), 3))
```

```python
for r in [0, 0.01, 0.1, 1, 5, 10, 100]:
    w0r, wr = train_linear_regression_reg(X_train, y_train, r)
    y_pred_r = w0r + X_val.dot(wr)
    print(r, round(rmse(y_val, y_pred_r), 4))
```

```python
scores = []
for seed in range(10):
    dtr, dva, dte = split(df, seed)
    dtr['horsepower'] = dtr['horsepower'].fillna(0)
    dva['horsepower'] = dva['horsepower'].fillna(0)
    Xtr, ytr = dtr[feat].values, dtr['fuel_efficiency_mpg'].values
    Xva, yva = dva[feat].values, dva['fuel_efficiency_mpg'].values
    w0s, ws = train_linear_regression(Xtr, ytr)
    ypred = w0s + Xva.dot(ws)
    scores.append(rmse(yva, ypred))

print(round(np.std(scores), 3))
```

```python
dtr, dva, dte = split(df, 9)
df_full_train = pd.concat([dtr, dva]).reset_index(drop=True)
df_full_train['horsepower'] = df_full_train['horsepower'].fillna(0)
dte['horsepower'] = dte['horsepower'].fillna(0)

X_full, y_full = df_full_train[feat].values, df_full_train['fuel_efficiency_mpg'].values
X_test, y_test = dte[feat].values, dte['fuel_efficiency_mpg'].values

w0f, wf = train_linear_regression_reg(X_full, y_full, r=0.001)
y_pred_test = w0f + X_test.dot(wf)
print(round(rmse(y_test, y_pred_test), 3))
```
