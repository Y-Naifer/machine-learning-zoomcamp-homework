```python
import numpy as np
```

```python
asia = df[df['origin'] == 'Asia']
asia.head()
```

```python
X = asia[['vehicle_weight', 'model_year']].iloc[:7].values
X
```

```python
XTX = X.T.dot(X)
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
w = XTX_inv.dot(X.T).dot(y)
w
```
